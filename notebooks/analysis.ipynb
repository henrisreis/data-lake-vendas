```python
import pandas as pd
import sys
from pathlib import Path
```

```python
PROJECT_DIR = Path.cwd()
sys.path.append(str(PROJECT_DIR / "src"))

from data_lake.bronze import dados_vendas
```

```python
data = pd.DataFrame(dados_vendas)
```

```python
data
```

```python
data.info()
```

```python
data = data.drop_duplicates(subset=['id_venda'])

data = data.dropna(how='any')

data = data[
    (data['quantidade'] > 0) & 
    (data['preco_unitario'] >= 0)
]

data['data_venda'] = pd.to_datetime(data['data_venda'], format='%Y-%m-%d')

data['id_venda'] = data['id_venda'].astype('int64')
data['quantidade'] = data['quantidade'].astype('int64')
data['preco_unitario'] = data['preco_unitario'].astype('float64')

data['valor_total'] = (
    data['quantidade'] * data['preco_unitario']
)
```
